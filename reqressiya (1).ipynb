```python
# Xətti Reqresiyya hesablanmasına nümunə 

import pandas as pd
import statsmodels.api as sm

# 1. Məlumat toplusunu (Data) yaradırıq
# X: Sosial media reklam büdcəsi (1.000 vahid)
# Y: Satış gəliri (1.000 vahid)
data = {
    'Reklam_Budcesi': [10, 15, 20, 25, 30, 35, 40, 45, 50, 55],
    'Satis_Geliri': [82, 98, 114, 131, 145, 162, 178, 193, 211, 226]
}

# Məlumatı Pandas DataFrame-ə çeviririk
df = pd.DataFrame(data)

# 2. Dəyişənləri təyin edirik
X = df['Reklam_Budcesi']
Y = df['Satis_Geliri']

# 3. Kəsişmə nöqtəsini (Intercept - Sabit) hesablamaq üçün X-ə sabit əlavə edirik
# Statsmodels avtomatik olaraq sabiti hesablamır, ona görə bu addım mütləqdir
X_sm = sm.add_constant(X)

# 4. OLS (Ordinary Least Squares - Ən Kiçik Kvadratlar) modelini qururuq
model = sm.OLS(Y, X_sm)
neticeler = model.fit()

# 5. Nəticələrin tam xülasəsini ekrana çap edirik
print(neticeler.summary())
```

```python
import pandas as pd
import statsmodels.api as sm
import numpy as np

# 1. NSDI Sistemi üçün Məlumat Toplusunu Yaradırıq
# X: Sisteme inteqrasiya olunan yeni istifadəçi/qurum sayı
# Y: 1 saat ərzində sistemə daxil olan API sorğularının sayı
data = {
    'Yeni_İstifadeci_Sayi': [0, 5, 10, 15, 20, 25, 30],
    'Saatliq_API_Sorgusu': [50, 75, 111, 166, 248, 369, 551]
}

df = pd.DataFrame(data)

# 2. Dəyişənləri təyin edirik
X = df['Yeni_İstifadeci_Sayi']
Y = df['Saatliq_API_Sorgusu']

# 3. Kəsişmə nöqtəsi (Sabit) əlavə edirik
X_sm = sm.add_constant(X)

# 4. Poisson Reqressiya Modelini Qururuq
# Xətti reqressiyada sm.OLS() istifadə edirdik, burada isə sm.Poisson() istifadə edirik
model = sm.Poisson(Y, X_sm)
neticeler = model.fit()

# 5. Nəticələrin xülasəsini çap edirik
print(neticeler.summary())

# 6. Əmsalın Eksponensialını (e^beta) hesablayırıq
beta_1 = neticeler.params['Yeni_İstifadeci_Sayi']
oran = np.exp(beta_1)

print("\n--- Xüsusi Hesablama ---")
print(f"Yeni İstifadəçi Əmsalı (Log-Odds): {beta_1:.4f}")
print(f"Eksponensial Dəyər (e^beta): {oran:.4f}")
```

```python
import pandas as pd
import statsmodels.api as sm
import numpy as np

# 1. Rəqəmsal İqtisadiyyat Analizi üçün Məlumat Toplusunu Yaradırıq
# X: Rəqəmsal ödəniş altyapısından istifadə faizi (%)
# Y: Vergi usulsüzlüyü edib-etməməsi (1 = Bəli, 0 = Xeyr)
data = {
    'Reqemsal_Odenis_Faizi': [10, 15, 20, 25, 30, 45, 50, 65, 80, 95],
    'Vergi_Usulsuzluyu':     [1,  1,  1,  1,  1,  0,  1,  0,  0,  0] 
}

df = pd.DataFrame(data)

# 2. Dəyişənləri təyin edirik
X = df['Reqemsal_Odenis_Faizi']
Y = df['Vergi_Usulsuzluyu']

# 3. Kəsişmə nöqtəsi (Sabit) əlavə edirik
X_sm = sm.add_constant(X)

# 4. Lojistik Reqressiya Modelini Qururuq
# Xətti reqressiyada OLS, Poisson-da Poisson, burada isə Logit funksiyasını çağırırıq
model = sm.Logit(Y, X_sm)
neticeler = model.fit()

# 5. Nəticələrin xülasəsini çap edirik
print(neticeler.summary())

# 6. Əmsalın Eksponensialını (Odds Ratio - Şans Oranını) hesablayırıq
beta_1 = neticeler.params['Reqemsal_Odenis_Faizi']
odds_ratio = np.exp(beta_1)
azalma_faizi = (1 - odds_ratio) * 100

print("\n--- Xüsusi Hesablama ---")
print(f"Rəqəmsal Ödəniş Əmsalı (Log-Odds): {beta_1:.4f}")
print(f"Şans Oranı (Odds Ratio - e^beta): {odds_ratio:.4f}")
print(f"Risk Azalma Faizi: {azalma_faizi:.2f}%")
```

```python
import pandas as pd
import numpy as np
from statsmodels.miscmodels.ordinal_model import OrderedModel

# 1. İqtisadi İslahatların Təhlili üçün Məlumat Toplusunu Yaradırıq
# X: Bölgəyə ayrılan təşviq büdcəsi (Milyon Manat)
# Y: Uğur Qavrayışı (1 = Düşük, 2 = Orta, 3 = Yüksək)
data = {
    'Tesvik_Butcesi': [2, 3, 4, 5, 7, 8, 9, 11, 12, 14, 15, 18],
    'Basari_Algisi':  [1, 1, 1, 2, 1, 2, 2,  3,  2,  3,  3,  3] 
}

df = pd.DataFrame(data)

# 2. Hədəf dəyişəninin "sıralı" (ordered) olduğunu sistemə mütləq bildirməliyik
df['Basari_Algisi'] = pd.Categorical(df['Basari_Algisi'], categories=[1, 2, 3], ordered=True)

# 3. Ordinal Lojistik Reqressiya Modelini Qururuq
# distr='logit' parametri bunun lojistik paylanma olduğunu bildirir
model = OrderedModel(df['Basari_Algisi'], df[['Tesvik_Butcesi']], distr='logit')

# Modelin optimizasiyası (bfgs metodu standart olaraq istifadə edilir)
neticeler = model.fit(method='bfgs')

# 4. Nəticələrin xülasəsini çap edirik
print(neticeler.summary())

# 5. Əmsalın Eksponensialını (Odds Ratio - Şans Oranını) hesablayırıq
beta_1 = neticeler.params['Tesvik_Butcesi']
odds_ratio = np.exp(beta_1)

print("\n--- Xüsusi Hesablama ---")
print(f"Təşviq Büdcəsi Əmsalı (Log-Odds): {beta_1:.4f}")
print(f"Sıçrama Şansı Oranı (e^beta): {odds_ratio:.4f}")
```

```python
import pandas as pd
import statsmodels.api as sm
import numpy as np

# 1. Rəqəmsal Ödəniş Növləri üçün Məlumat Toplusunu Yaradırıq
# Y: 0 = Bank Kartı (Baza), 1 = Elektron Pul Kisəsi, 2 = Kriptovalyuta
data = {
    'Ayliq_Gelir': [1.0, 1.5, 2.0, 2.5, 3.5, 4.0, 5.0, 6.0, 7.0, 8.5],
    'Odenis_Usulu': [0, 0, 0, 1, 1, 1, 2, 2, 2, 2] 
}

df = pd.DataFrame(data)

# 2. Dəyişənləri təyin edirik
X = df['Ayliq_Gelir']
Y = df['Odenis_Usulu']

# 3. Kəsişmə nöqtəsi əlavə edirik
X_sm = sm.add_constant(X)

# 4. Multinominal Lojistik Reqressiya Modelini Qururuq
model = sm.MNLogit(Y, X_sm)
neticeler = model.fit()

# 5. Nəticələrin xülasəsini çap edirik
print(neticeler.summary())

# 6. Əmsalların Eksponensialını (Şans Oranlarını) hesablayırıq
# Cədvəldə hər bir kateqoriya üçün ayrı tənlik (əmsal) çıxacaq
beta_1 = neticeler.params.iloc[1, 0] # E-pul kisəsi üçün gəlir əmsalı
beta_2 = neticeler.params.iloc[1, 1] # Kripto üçün gəlir əmsalı

odds_ratio_1 = np.exp(beta_1)
odds_ratio_2 = np.exp(beta_2)

print("\n--- Xüsusi Hesablama ---")
print(f"Elektron Pul Kisəsi seçmə Şans Oranı (Kart ilə müqayisədə): {odds_ratio_1:.2f}")
print(f"Kriptovalyuta seçmə Şans Oranı (Kart ilə müqayisədə): {odds_ratio_2:.2f}")
```
