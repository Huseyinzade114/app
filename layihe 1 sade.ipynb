```python
"""
Ssenari: Tədris Metodlarının Müqayisəsi
Universitetdə bir fənn üzrə iki fərqli tədris metodu tətbiq olunur. Metod_A ənənəvi mühazirə üsuludur, Metod_B isə yeni interaktiv öyrənmə metodudur. 
Hər qrupda 30 tələbə var. Semestrin sonunda tələbələrin imtahan nəticələri (100 ballıq sistem) qeyd olunub. 
Tədqiqatın məqsədi yeni metodun (Metod_B) ənənəvi metodla (Metod_A) müqayisədə tələbə nailiyyətlərinə fərqli təsir edib-etmədiyini yoxlamaqdır.
"""    
```

```python
import pandas as pd
from scipy import stats
import numpy as np
```

```python
# A qrupu üçün nəticəni gətirək 

df = pd.read_excel(r"C:\Users\ACER\Desktop\Layihə.xlsx",index_col="Telebe_ID")
df.head(5)
```

```python
# A qrupu üçün nəticəni gətirək 
df2 = pd.read_excel(r"C:\Users\ACER\Desktop\Layihə.xlsx",index_col="Telebe_ID",sheet_name="B")
df2.head(5)
```

```python
# her bir qrup üçün ayrılıqda normal paylanmanı yoxlayaq 
# A qrupu üçün Shapiro-Wilk testi (df)
stat_a, p_value_a = stats.shapiro(df['Imtahan_Neticesi'].dropna())

print("--- A Qrupu ---")
print(f"Test Statistikası: {stat_a:.4f}")
print(f"p-dəyəri: {p_value_a:.4f}")

if p_value_a > 0.05:
    print("Nəticə: Data normal paylanıb (H0 rədd edilmir).")
else:
    print("Nəticə: Data normal paylanmayıb (H0 rədd edilir).")

print("\n========================\n")

# B qrupu üçün Shapiro-Wilk testi (df2)
stat_b, p_value_b = stats.shapiro(df2['Imtahan_Neticesi'].dropna())

print("--- B Qrupu ---")
print(f"Test Statistikası: {stat_b:.4f}")
print(f"p-dəyəri: {p_value_b:.4f}")

if p_value_b > 0.05:
    print("Nəticə: Data normal paylanıb (H0 rədd edilmir).")
else:
    print("Nəticə: Data normal paylanmayıb (H0 rədd edilir).")
```

```python

# A və B qruplarının imtahan nəticələri üçün Levene testinin aparılması
stat_levene, p_value_levene = stats.levene(df['Imtahan_Neticesi'].dropna(), df2['Imtahan_Neticesi'].dropna())

print("--- Levene Testi (Variansların Bərabərliyi) ---")
print(f"Test Statistikası: {stat_levene:.4f}")
print(f"p-dəyəri: {p_value_levene:.4f}")

if p_value_levene > 0.05:
    print("Nəticə: Varianslar bərabərdir (H0 rədd edilmir).")
    print("Əla! Həm normallıq, həm də variansların bərabərliyi şərtləri ödəndiyi üçün standart 'Independent t-test' apara bilərik.")
else:
    print("Nəticə: Varianslar bərabər deyil (H0 rədd edilir).")
    print("Bu halda Welch t-testi (equal_var=False) aparmaq lazım gələcək.")
```

```python


# Müstəqil Seçmələr üçün t-test (equal_var=True edirik, çünki Levene testindən keçdik)
stat_t, p_value_t = stats.ttest_ind(df['Imtahan_Neticesi'].dropna(), df2['Imtahan_Neticesi'].dropna(), equal_var=True)

print("--- Müstəqil Seçmələr üçün T-Test Nəticəsi ---")
print(f"Test Statistikası (t-value): {stat_t:.4f}")
print(f"p-dəyəri: {p_value_t:.4f}\n")

print("--- Nəticənin Şərhi (Tədris Metodları Ssenarisi) ---")
if p_value_t > 0.05:
    print("Qərar: H0 rədd edilmir (p > 0.05).")
    print("Yekun: Yeni interaktiv öyrənmə metodunun (Metod_B) ənənəvi mühazirə üsulu ilə (Metod_A) müqayisədə tələbələrin imtahan nəticələrinə fərqli bir təsiri yoxdur. Hər iki metod oxşar nəticələr verir.")
else:
    print("Qərar: H0 rədd edilir (p <= 0.05).")
    print("Yekun: Yeni interaktiv öyrənmə metodu (Metod_B) ilə ənənəvi mühazirə üsulu (Metod_A) arasında tələbə nailiyyətlərinə təsir baxımından statistik olaraq əhəmiyyətli fərq VAR.")
    
    # Fərqin kimin xeyrinə olduğunu tapmaq üçün ortalamaları hesablayaq
    mean_A = df['Imtahan_Neticesi'].mean()
    mean_B = df2['Imtahan_Neticesi'].mean()
    
    print(f"\nƏlavə Məlumat - Ortalamalar:")
    print(f"Metod_A (Ənənəvi) Ortalama: {mean_A:.2f}")
    print(f"Metod_B (İnteraktiv) Ortalama: {mean_B:.2f}")
    
    if mean_B > mean_A:
        print("Nəticə: İnteraktiv metod (Metod_B) ənənəvi metoddan DAHA EFFEKTİVDİR.")
    else:
        print("Nəticə: Ənənəvi metod (Metod_A) interaktiv metoddan DAHA EFFEKTİVDİR.")
```
