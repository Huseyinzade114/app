```python
# Python-da K-S testi etmək üçün scipy.stats kitabxanasından istifadə edirik.

"""
Vacib Məqam: kstest funksiyası susmaya görə (default) verilənləri Standart Normal Paylanma (Mean=0, Std=1) ilə müqayisə edir. 
Sizin verilənlərinizin ortalaması fərqli ola biləcəyi üçün, funksiyaya verilənlərinizin 
ortalamasını və standart kənarlaşmasını arqument kimi ötürməlisiniz.
""" 
```

```python
import numpy as np
from scipy import stats

# 1. Nümunə verilənlər yaradaq (Normal paylanmaya uyğun)
np.random.seed(42)
data = np.random.normal(loc=115, scale=10, size=1000)

# 2. Verilənlərin parametrlərini hesablayaq (Bu addım K-S üçün vacibdir!)
mean_val = np.mean(data)
std_val = np.std(data, ddof=1)

# 3. K-S Testini icra edək
# 'norm' - normal paylanma ilə müqayisə etmək üçün
# args=(mean, std) - bizim verilənlərin parametrlərinə uyğunlaşdırmaq üçün
d_statistic, p_value = stats.kstest(data, 'norm', args=(mean_val, std_val))

print(f"K-S Statistikası (D): {d_statistic:.4f}")
print(f"P-dəyəri: {p_value:.4f}")

# 4. Nəticənin avtomatik izahı
alpha = 0.05
if p_value > alpha:
    print("Nəticə: Paylanma Normaldır (H0 rədd edilmir)")
else:
    print("Nəticə: Paylanma Normal DEYİL (H0 rədd edilir)")
```

```python
# iki numuneli K-S testi 

# 1. İki fərqli nümunə (sample) yaradaq
# A qrupu: Ortalama 50 olan paylanma
qrup_A = np.random.normal(loc=100, scale=10, size=500)

# B qrupu: Ortalama 52 olan paylanma (A-ya çox yaxındır, görək fərqi tutacaqmı?)
qrup_B = np.random.normal(loc=52, scale=10, size=500)

# 2. İki Nümunəli K-S Testini İcra Edək
# Burada artıq 'norm' yazmağa ehtiyac yoxdur, birbaşa iki datanı veririk
ks_statistic, p_value = stats.ks_2samp(qrup_A, qrup_B)

print(f"K-S Statistikası (D məsafəsi): {ks_statistic:.4f}")
print(f"P-dəyəri: {p_value:.4f}")

if p_value > 0.05:
    print("Nəticə: İki qrupun paylanması EYNİDİR (H0 rədd edilmir).")
else:
    print("Nəticə: İki qrupun paylanması FƏRQLİDİR (H0 rədd edilir).")
```
