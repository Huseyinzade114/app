```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
                                                                   #Levene Testi və Avtomatik Qərar
# 1. Nümunə Verilənlərin Yaradılması (Simulyasiya)
# A Qrupu: Ortalama=50, Standart Kənaraçıxma=5 (Stabil qrup)
group_A = np.random.normal(loc=50, scale=5, size=30)

# B Qrupu: Ortalama=55, Standart Kənaraçıxma=15 (Dəyişkən qrup - Varyans fərqlidir!)
group_B = np.random.normal(loc=55, scale=15, size=30)

# 2. Levene Testinin Həyata Keçirilməsi
# center='median' (varsayılan) ən məsləhətlisidir. Orijinal düstur üçün 'mean' yaza bilərsiniz.
stat_w, p_value = stats.levene(group_A, group_B, center='median')

# 3. Nəticələrin Çap Edilməsi
print(f"Levene Statistikası (W): {stat_w:.4f}")
print(f"P-dəyəri (Sig.): {p_value:.4f}")

print("-" * 30)

# 4. Avtomatik Qərar Məntiqi (Tələbələr üçün)
alpha = 0.05

if p_value > alpha:
    print("NƏTİCƏ: P > 0.05")
    print("✅ Varyanslar Bərabərdir (Homogendir).")
    print("➔ Növbəti addım: 'Standart T-testi' istifadə edin (equal_var=True).")
else:
    print("NƏTİCƏ: P < 0.05")
    print("❌ Varyanslar Bərabər Deyil (Homogenlik pozuldu).")
    print("➔ Növbəti addım: 'Welch T-testi' istifadə edin (equal_var=False).")
```

```python
# One-Sample T-test (Bir Seçməli T-testi) 

# 1. Verilənlərin Yaradılması (Simulyasiya)
# Tutaq ki, bizim ölçdüyümüz real paketlər orta hesabla 495 qram gəlir (biraz əskik)
# np.random.normal(ortalama, standart_kənaraçıxma, say)
np.random.seed(42) # Nəticə hər dəfə eyni olsun deyə
sample_data = np.random.normal(loc=498, scale=5, size=30)

# 2. Testin Parametrləri
expected_mean = 500  # H0: Populyasiya ortalaması (İddia edilən)

# 3. One-Sample T-testinin Həyata Keçirilməsi
# null_hypothesis: sample_mean == expected_mean
t_stat, p_value = stats.ttest_1samp(sample_data, popmean=expected_mean)

# 4. Nəticələrin Çap Edilməsi
print(f"Seçmənin Ortalaması (Real): {np.mean(sample_data):.2f}")
print(f"Gözlənilən Ortalama (H0): {expected_mean}")
print("-" * 30)
print(f"T-statistikası: {t_stat:.4f}")
print(f"P-dəyəri: {p_value:.4f}")
print("-" * 30)

# 5. Qərarın Verilməsi
alpha = 0.05

if p_value < alpha:
    print("NƏTİCƏ: H0 Rədd edilir (P < 0.05).")
    print(f"Statistik olaraq əhəmiyyətli fərq var! Paketlər {expected_mean} qram deyil.")
else:
    print("NƏTİCƏ: H0 Qəbul edilir (P > 0.05).")
    print(f"Fərq təsadüfidir. Paketlərin {expected_mean} qram olmadığını demək üçün sübut yoxdur.")
```

```python
# Independent Samples T-test (bagimsiz)

# 1. Verilənlər (İki ayrı qrup)
qrup_A = [80, 85, 78, 90, 82, 75, 88, 92]
qrup_B = [70, 75, 72, 85, 78, 65, 80, 82]

# 2. Testin hesablanması
# equal_var=True (Varyanslar bərabərdirsə)
# equal_var=False (Varyanslar fərqlidirsə - Welch)
t_stat, p_value = stats.ttest_ind(qrup_A, qrup_B, equal_var=True)

# 3. Nəticə
print(f"T-statistikası: {t_stat}")
print(f"P-dəyəri: {p_value}")
```

```python
# Cütlənmiş (Paired) T-test üçün Python-da scipy.stats kitabxanasının ttest_rel funksiyası istifadə olunur ("rel" = related, yəni əlaqəli deməkdir).


# 1. Əlimizdə olan iki data (Məsələn: Əvvəl və Sonra)
# Qeyd: Hər iki siyahıda eyni sayda rəqəm olmalıdır!
data_evvel = [80, 85, 78, 90, 82, 75, 88, 92]
data_sonra = [85, 87, 80, 94, 85, 79, 90, 95]

# 2. Testin icrası (Paired T-Test)
t_stat, p_value = stats.ttest_rel(data_evvel, data_sonra)

# 3. Nəticənin ekrana çıxarılması
print(f"T-statistikası: {t_stat:.4f}")
print(f"P-dəyəri: {p_value:.4f}")
```

```python
# elave numune

# 1. Random Verilənlərin Yaradılması
np.random.seed(42) # Nəticə hər dəfə eyni olsun deyə
# Ortalama=50, Standart Kənaraçıxma=10, Say=30
qrup_A = np.random.normal(loc=50, scale=10, size=30)
# Ortalama=55, Standart Kənaraçıxma=10, Say=30 (Biraz fərqli yaratdıq)
qrup_B = np.random.normal(loc=55, scale=10, size=30)

# 2. Testin Hesablanması (Independent)
# equal_var=True (Varyansların bərabər olduğunu fərz edirik)
t_stat, p_value = stats.ttest_ind(qrup_A, qrup_B, equal_var=True)

# 3. Nəticələrin Çapı
print(f"T-statistikası: {t_stat:.4f}")
print(f"P-dəyəri: {p_value:.4f}")
print("-" * 30)

# 4. Hipotezin Yoxlanması (Qərar)
alpha = 0.05 # Qəbul edilən xəta payı (95% əminlik)

if p_value < alpha:
    print("NƏTİCƏ: H0 Rədd edilir.")
    print("✅ İki qrup arasında statistik olaraq əhəmiyyətli FƏRQ VAR.")
else:
    print("NƏTİCƏ: H0 Qəbul edilir (Rədd edilə bilmir).")
    print("❌ İki qrup arasında statistik fərq YOXDUR.")
```

```python
# 1. Random Verilənlərin Yaradılması
np.random.seed(42)
# Əvvəlki vəziyyət
evvel = np.random.normal(loc=60, scale=5, size=30)
# Sonrakı vəziyyət (Əvvəlkinin üzərinə biraz artım gəlirik ki, fərq yaransın)
sonra = evvel + np.random.normal(loc=2, scale=1, size=30)

# 2. Testin Hesablanması (Paired - Əlaqəli)
t_stat, p_value = stats.ttest_rel(evvel, sonra)

# 3. Nəticələrin Çapı
print(f"T-statistikası: {t_stat:.4f}")
print(f"P-dəyəri: {p_value:.4f}")
print("-" * 30)

# 4. Hipotezin Yoxlanması (Qərar)
alpha = 0.05

if p_value < alpha:
    print("NƏTİCƏ: H0 Rədd edilir.")
    print("✅ Əvvəl və Sonra arasında statistik olaraq əhəmiyyətli FƏRQ VAR.")
else:
    print("NƏTİCƏ: H0 Qəbul edilir (Rədd edilə bilmir).")
    print("❌ Əvvəl ve Sonra arasında statistik fərq YOXDUR.")
```
