```python
import pandas as pd
import scipy.stats as stats
import numpy as np
```

```python
df = pd.read_excel(r"C:\Users\ACER\Desktop\Data Analitika\ANOVA test.xlsx")
df
```

```python


print("--- ADDIM 1: NORMALLIQ (SHAPIRO) ---")
# P > 0.05 olmalıdır
print(f"Eyani P-dəyəri: {stats.shapiro(df['Metod_A']).pvalue:.4f}")
print(f"Online P-dəyəri: {stats.shapiro(df['Metod_B']).pvalue:.4f}")
print(f"Hibrid P-dəyəri: {stats.shapiro(df['Metod_C']).pvalue:.4f}")
```

```python
print("\n--- ADDIM 2: HOMOGENLİK (LEVENE) ---")
# P > 0.05 olmalıdır (Varyanslar bərabərdirmi?)
stat, p_levene = stats.levene(df['Metod_A'], df['Metod_B'], df['Metod_C'])
print(f"Levene P-dəyəri: {p_levene:.4f}")
```

```python
print("\n--- ADDIM 3: ANOVA TESTİ ---")
# P < 0.05 olmalıdır (Fərq varmı?)
f_stat, p_anova = stats.f_oneway(df['Metod_A'], df['Metod_B'], df['Metod_C'])
print(f"ANOVA P-dəyəri: {p_anova:.4f}") 

if p_anova < 0.05:
    print("\nNƏTİCƏ: Fərq var! (H0 rədd edilir)")
else:
    print("netice ferqi yoxdur")
```

```python

# Datanı Tukey testinin başa düşəcəyi formata salırıq (Melt edirik)
df_melt = pd.melt(df, value_vars=['Metod_A', 'Metod_B', 'Metod_C'], 
                  var_name='Metodlar', value_name='Qiymetler')

# Əgər datanızda boş (NaN) dəyərlər varsa, xəta verməməsi üçün onları silirik
df_melt = df_melt.dropna()
df_melt
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Tukey HSD testini tətbiq edirik
tukey_netice = pairwise_tukeyhsd(endog=df_melt['Qiymetler'], 
                                 groups=df_melt['Metodlar'], 
                                 alpha=0.05)

# Nəticə cədvəlini ekrana yazdırırıq
print("\n--- ADDIM 4: TUKEY HSD TESTİ ---")
print(tukey_netice)
```
