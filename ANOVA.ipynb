```python
# Data var -  Normallığı yoxla (Rəqəmlə) - 2. Homogenliyi yoxla (Rəqəmlə) - 3. ANOVA testini et.
```

```python
import pandas as pd
import scipy.stats as stats


# --- 1. DATANIN YARADILMASI (Sanki Excel-dən oxuyuruq) ---
data = {
    'Metod': (['Eyani'] * 10) + (['Online'] * 10) + (['Hibrid'] * 10),
    'Xal': [
        75, 78, 72, 79, 74, 76, 73, 77, 75, 71,  # Eyani (Stabil, orta)
        60, 65, 62, 68, 64, 66, 63, 67, 61, 59,  # Online (Aşağı nəticələr)
        85, 88, 82, 89, 84, 86, 83, 87, 85, 81   # Hibrid (Yüksək nəticələr)
    ]
}

df = pd.DataFrame(data)

# Qrupları ayırırıq (Analiz üçün)
eyani = df[df['Metod'] == 'Eyani']['Xal']
online = df[df['Metod'] == 'Online']['Xal']
hibrid = df[df['Metod'] == 'Hibrid']['Xal']

print("--- ADDIM 1: NORMALLIQ (SHAPIRO) ---")
# P > 0.05 olmalıdır
print(f"Eyani P-dəyəri: {stats.shapiro(eyani).pvalue:.4f}")
print(f"Online P-dəyəri: {stats.shapiro(online).pvalue:.4f}")
print(f"Hibrid P-dəyəri: {stats.shapiro(hibrid).pvalue:.4f}")

print("\n--- ADDIM 2: HOMOGENLİK (LEVENE) ---")
# P > 0.05 olmalıdır (Varyanslar bərabərdirmi?)
stat, p_levene = stats.levene(eyani, online, hibrid)
print(f"Levene P-dəyəri: {p_levene:.4f}")

print("\n--- ADDIM 3: ANOVA TESTİ ---")
# P < 0.05 olmalıdır (Fərq varmı?)
f_stat, p_anova = stats.f_oneway(eyani, online, hibrid)
print(f"ANOVA P-dəyəri: {p_anova:.10f}") 

if p_anova < 0.05:
    print("\nNƏTİCƏ: Fərq var! (H0 rədd edilir)")
else:
    print("netice ferqi yoxdur")
```
